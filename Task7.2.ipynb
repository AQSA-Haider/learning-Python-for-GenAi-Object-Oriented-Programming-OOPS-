# Modifying the Product class
# Making price a private attribute-using double underscore
# Creating getter and setter method:
# ~get_price()
# ~set_price(new_price).


```python
class Product:
    def __init__(self,name,price ,category):
        self.name=name
        self.__price=price ## Now price has become a private attribute because if double underscore
        self.category=category

    def get_price(self):
       return self.__price
    
    def set_price(self, new_price):
       if new_price > 0:
         self.__price = new_price
         print(f"Price updated to {self.__price:.2f}Rs.")
       else:
        print("Invalid price. Price must be greater than 0.")
    
    def get_info(self):
       print(f"Product: {self.name}")
```

```python
item = Product("Headphone","1500", "Audio")
item.get_info()
```

# Testing modified prices using set_price(new_price) -only update if the price is > 0..

```python
item.set_price(1300.99)
print(f"Current Price via getter: {item.get_price():.2f}")
```

```python
item.set_price(-56)
```
