# Polymorphism
# Creating two classes:
# ~Laptop(product)
# ~Mobile(product)

```python
class Product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
    
    def get_info(self):
        print(f"The product {self.name} is of {self.price} Rs.")
```

# Both override get_info()

```python
class Laptop(Product):
    def __init__(self, name, price, ram):
        super().__init__(name,price)
        self.ram = ram
    def get_info(self):
        return f"[Laptop] {self.name} ({self.ram} Ram) - {self.price}Rs."

class Mobile(Product):
    def __init__(self, name, price, battery):
        super().__init__(name,price)
        self.battery = battery
    def get_info(self):
        return f"[Mobile] {self.name} ({self.battery} mAh Battery) - {self.price}Rs."

    
```

# Writing a loop that iterates over objects of Laptop and Mobile and calls get_info() on each to show polymorphism.

```python
devices = [Laptop("Dell XPS", 100000, "256GB"), Mobile("Samsung Galaxy", 75000, 10000)]
for item in devices:
    print(item.get_info())
```
