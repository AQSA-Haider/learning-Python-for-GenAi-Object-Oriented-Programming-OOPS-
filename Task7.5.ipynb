# Abstraction (Using Abstract Base Class)
# Creating an Abstract class Payment with abstract method:
# process_payment(amount)

```python
from abc import ABC,abstractmethod
class Payment(ABC):
    @abstractmethod
    def process_payment(self,amount):
        pass    
```

# Creating the first subclass
# CreditCardPayment()

```python
class CreditCardPayment(Payment):
    def process_payment(self, amount):
        print(f"Processing Credit Card payment of Rs.{amount}.")
```

# Creating the second subclass 
# UPIPayment()

```python
class UPIPayment(Payment):
    def process_payment(self,amount):
        print(f"Processing UPI payment of Rs.{amount}.")
```

# Both subclasses are overriding process_payment()method with print statement.
# Finally, testing all the classes.

```python
c_pay = CreditCardPayment()
upi_pay = UPIPayment()
 
c_pay.process_payment(500000)
upi_pay.process_payment(1000)
```
