# Basic Class and Object Creation
#  Creating a class Product
# Attributes: name,price,category
# Methods: get_info--> prints the product details
# Optional: creating one more method apply_discount(Percent) that returns the discounted price.

```python
class Product:
    def __init__(self,name,price ,category):
        self.name=name
        self.price=price
        self.category=category
    
    def get_info(self):
        print(f"The product {self.name} is of {self.price} rs. and is from {self.category} category.")
    
    def apply_discount(self,percent):
        return (self.price-self.price*percent/100)

```

# creating the first method - product1

```python
product1=Product("sneakers",2000,"shoes")
print(product1.get_info())
print(product1.apply_discount(20))
```

# creating the second method - product2

```python
product2=Product("lip tint",500,"makeup")
print(product2.get_info())
print(product2.apply_discount(20))
```
