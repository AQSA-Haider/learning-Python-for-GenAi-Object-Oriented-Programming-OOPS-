# Here we have our initial Product class:
# and we have added __str__ mathod 
# and it Returns a readable string:
# Product(name, price, category) 
# operator overloading(__add__)
# allows product1 + product2
# to return the total combined price.

```python
class Product:
    def __init__(self,name,price ,category):
        self.name=name
        self.price=price
        self.category=category

    def __str__(self) -> str:
        return  f"Product( {self.name}, {self.price}, {self.category})"
    
    def __add__(self, other):
        if isinstance(other, Product):
            return self.price + other.price
        return NotImplemented
    
```

# Testing this two product objects. 

```python
product1 = Product("Laptop",75000, "Electronics")
product2 = Product("Headphones", 3000, "Accessories")

print(product1)
print(product2)
```

```python
total_price = product1 + product2
print(f"Total Combined Price: {total_price}")
```
